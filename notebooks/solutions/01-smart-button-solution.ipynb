## 参考 C++ 与行为检查

`was_pressed_` 保存上一份输入，`initialized_` 建立首次采样 baseline。只有从“没按下”变为“按下”才改变 LED。


```python
%%cpp_lab solution exercise pass
#pragma once
#include "academy/board.hpp"

namespace academy {
class Controller {
public:
    explicit Controller(Board& board) : board_(board) {}

    void tick() {
        const bool pressed = board_.read_button() == Level::low;
        if (!initialized_) {
            was_pressed_ = pressed;
            initialized_ = true;
        } else if (pressed && !was_pressed_) {
            led_on_ = !led_on_;
        }
        was_pressed_ = pressed;
        board_.write_led(led_on_);
    }

private:
    Board& board_;
    bool led_on_ = false;
    bool was_pressed_ = false;
    bool initialized_ = false;
};
} // namespace academy
```
